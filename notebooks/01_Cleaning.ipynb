```python
import pandas as pd

# 1. Accounts
df = pd.read_csv("ravenstack_accounts.csv")

boolean_cols = [
    "is_trial",
    "churn_flag"
]

for col in boolean_cols:
    df[col] = df[col].map({
        True: 1,
        False: 0,
        "True": 1,
        "False": 0
    })

df.to_csv("ravenstack_accounts_clean.csv", index=False)

print("✅ Accounts file cleaned successfully!")
```

```python
import pandas as pd

df = pd.read_csv("ravenstack_subscriptions.csv")

boolean_cols = [
    "is_trial",
    "upgrade_flag",
    "downgrade_flag",
    "churn_flag",
    "auto_renew_flag"
]

for col in boolean_cols:
    df[col] = df[col].map({True: 1, False: 0})

df.to_csv("subscriptions_clean.csv", index=False)
```

```python
import pandas as pd

df = pd.read_csv("subscriptions_clean.csv")

df["start_date"] = pd.to_datetime(df["start_date"]).dt.strftime("%Y-%m-%d")
df["end_date"] = pd.to_datetime(df["end_date"], errors="coerce").dt.strftime("%Y-%m-%d")

df.to_csv("ravenstack_subscriptions_clean.csv", index=False)

print("✅ Dates cleaned and file saved!")
```

```python
# 3. Feature Usage
# Load original CSV
df = pd.read_csv("ravenstack_feature_usage.csv")

# Convert Boolean → 1/0
df["is_beta_feature"] = df["is_beta_feature"].map({
    True: 1,
    False: 0,
    "True": 1,
    "False": 0
})

# Convert date
df["usage_date"] = pd.to_datetime(df["usage_date"]).dt.strftime("%Y-%m-%d")

# Save cleaned file
df.to_csv("ravenstack_feature_usage_clean.csv", index=False)

print("✅ Feature usage cleaned successfully!")
print(df.head())
```

```python
# 4. Support Tickets
df = pd.read_csv("ravenstack_support_tickets.csv")

# Convert Boolean → 1/0
df["escalation_flag"] = df["escalation_flag"].map({
    True: 1,
    False: 0,
    "True": 1,
    "False": 0
})

# Convert dates
df["submitted_at"] = pd.to_datetime(
    df["submitted_at"],
    errors="coerce"
).dt.strftime("%Y-%m-%d")

df["closed_at"] = pd.to_datetime(
    df["closed_at"],
    errors="coerce"
).dt.strftime("%Y-%m-%d")

# Save cleaned CSV
df.to_csv("ravenstack_support_tickets_clean.csv", index=False)

print("✅ Support tickets cleaned!")
print(df.head())
```

```python

# 5. Churn Events
import pandas as pd

df = pd.read_csv("ravenstack_churn_events.csv")

boolean_cols = [
    "preceding_upgrade_flag",
    "preceding_downgrade_flag",
    "is_reactivation"
]

for col in boolean_cols:
    df[col] = df[col].map({
        True: 1,
        False: 0,
        "True": 1,
        "False": 0
    })

df["churn_date"] = pd.to_datetime(
    df["churn_date"],
    errors="coerce"
).dt.strftime("%Y-%m-%d")

df.to_csv("ravenstack_churn_events_clean.csv", index=False)

print("✅ Churn events cleaned!")
print(df.head())

print("✅ All 3 files cleaned successfully!")
```

```python
accounts = pd.read_csv("ravenstack_accounts_clean.csv")
subscriptions = pd.read_csv("ravenstack_subscriptions_clean.csv")
feature_usage = pd.read_csv("ravenstack_feature_usage_clean.csv")
support_tickets = pd.read_csv("ravenstack_support_tickets_clean.csv")
churn_events = pd.read_csv("ravenstack_churn_events_clean.csv")

print("✅ All datasets loaded")
```

```python
datasets = {
    "Accounts": accounts,
    "Subscriptions": subscriptions,
    "Feature Usage": feature_usage,
    "Support Tickets": support_tickets,
    "Churn Events": churn_events
}

for name, df in datasets.items():
    print(f"\n{name}")
    print("Shape:", df.shape)
    print("Duplicates:", df.duplicated().sum())
    print("Missing values:", df.isna().sum().sum())
```

```python
for name, df in datasets.items():
    missing = df.isna().sum()
    missing = missing[missing > 0]

    if len(missing) > 0:
        print(f"\n{name}")
        print(missing)
```
